Ensemble Learning is an ML technique that combines several models to create an optimal prediction model. There are several types of Ensemble Learning, being the most common Max Voting (Majority Voting), BAGGing (Boostrap AGGregating), Boosting and Stacking:
- **Max** Voting involves fitting several models, each of which makes a prediction and votes on each sample. Only the class with the most votes is included in the final prediction;
- **Bagging** involves adjusting decision trees to different samples from the same dataset and calculating the average of these predictions;
- **Boosting** involves sequentially adding ensemble models that correct the predictions made by previous models, producing a weighted average of the predictions;
- **Stacking** involves fitting different types of models to the same dataset and using another model to determine the optimal combinations of predictions.
